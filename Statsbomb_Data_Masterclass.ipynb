```python
import pandas as pd
from mplsoccer import Sbopen
```

```python
Sbopen().competition()
```

#### Competitions

```python
df_competitions = Sbopen().competition()
```

```python
df_competitions
```

```python
df_competitions.to_csv('Statsbomb_Competitions.csv', index=False)
```

#### Matches

```python
df_matches = Sbopen().match(competition_id = 53, season_id = 315)
```

```python
df_matches
```

```python
df_matches.to_csv('Statsbomb_Matches.csv', index=False)
```

#### Events

```python
df_events = Sbopen().event(4020846)[0] # Only event data
```

```python
df_events
```

```python
df_events.to_csv('Statsbomb_Events.csv', index=False)
```
